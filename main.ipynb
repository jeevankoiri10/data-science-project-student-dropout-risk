````python
import pandas as pd 

# Show all columns for the rest of the session
pd.set_option("display.max_columns", None)

# Load the dataset
df = pd.read_csv('student_dropout_risk_2026.csv')
df.head(10)
````

````python
# check whether there are any missing values in the dataframe
df.notna().all().all() # True means no missing values
````

### 1. Extract all students that study Online, have an attendance_rate smaller than 50, and have more than 5 missed_deadlines. Store them in a dataframe called at_risk

````python
at_risk = df[(df['study_mode'] == 'Online') & (df['attendance_rate'] < 50) & (df['missed_deadlines'] > 5)]
at_risk
````

### 2. Then compute, for each `course_type`, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called `at_risk_by_course`, with columns `students` and `dropout_rate`, sorted by decreasing `dropout_rate`.

````python
at_risk_by_course = (
    at_risk.groupby("course_type")
      .agg(
          students=("person_id", "count"),      # or ("dropped_out", "size") to include NaNs
          dropout_rate=("dropout", "mean"),  # mean of 0/1 indicator = fraction dropped out
      )
      .sort_values("dropout_rate", ascending=False)
)
at_risk_by_course.head(10)
````

### 3. Add to the dataframe the following columns:
- `pass_rate`: the ratio between `units_passed` and `units_enrolled`;
- `gpa_change`: the difference between `gpa_semester_2` and `gpa_semester_1`;
- `gpa_trend`: a string that is `"improving"` if `gpa_change` is larger than 0.5, `"declining"` if `gpa_change` is smaller than -0.5, and `"stable"` otherwise.

````python
df['pass_rate'] = df['units_passed'] / df['units_enrolled']
df['gpa_change'] = df['gpa_semester_2'] - df['gpa_semester_1']
df['gpa_trend'] = df['gpa_change'].apply(lambda x: "improving" if x > 0.5 else ("declining" if x < -0.5 else "stable"))   

df 
````

### 4. Verify that, for every student, `units_passed + units_failed` equals `units_enrolled`; if not, report the rows violating this condition.


````python
violations = df[df["units_passed"] + df["units_failed"] != df["units_enrolled"]]

print(type(violations))

if violations.empty:
    print("All rows satisfy units_passed + units_failed == units_enrolled.")
else:
    print(f"Found {len(violations)} violating row(s):")
    print(violations)
````

### 5. Compute the dropout rate for each value of `gpa_trend`, and the `person_id` of the 10 students with the largest GPA drop.

````python

# the dropout rate for each value of gpa_trend
dropout_by_trend = (
    df.groupby("gpa_trend")["dropout"]
      .mean()
      .reset_index(name="dropout_rate")
      .sort_values("dropout_rate", ascending=False)
)

print("\nDropout rate by GPA trend:")
print(dropout_by_trend)


# the `person_id` of the 10 students with the largest GPA drop
largest_gpa_drops = df.nsmallest(10, "gpa_change")[["person_id", "gpa_change", "gpa_trend"]]
top_10_drop_ids = largest_gpa_drops["person_id"]
largest_gpa_drops

print("\nTop 10 students with the largest GPA drop:")
print(largest_gpa_drops)
print(top_10_drop_ids.tolist())  # as a plain list of IDs
````

### 6. Partition the students into the following groups, according to `weekly_work_hours`: `0`, `1-10`, `11-20`, `21-30`, `more than 30` (all intervals include both extremes). 

````python
import numpy as np

df["work_hours_group"] = pd.cut(
    df["weekly_work_hours"],
    bins=[-np.inf, 0, 10, 20, 30, np.inf],
    labels=["0", "1-10", "11-20", "21-30", "more than 30"],
    right=True,
)

df[["person_id", "weekly_work_hours", "work_hours_group"]].head()
````

````python
# to check correct numbers are being placed in a correct group
df.loc[
    df["weekly_work_hours"].isin([0, 1, 10, 11, 20, 21, 30, 31]),
    ["weekly_work_hours", "work_hours_group"],
].drop_duplicates().sort_values("weekly_work_hours")
````

### 7. Build a risk score by summing:

- 2 points if the tuition is not up to date;
- 1 point if `stress_level` is at least 8;
- 1 point if `sleep_hours` is smaller than 5;o
- 1 point if the student has a part-time job with more than 20 weekly work hours;
- 2 points if `attendance_rate` is smaller than 60;
- 1 point if `missed_deadlines` is larger than 3.

### For each value of the risk score, compute the number of students and the dropout rate.

````python
tuition_risk = (df["tuition_up_to_date"] == 0) * 2
stress_risk = (df["stress_level"] >= 8) * 1
sleep_risk = (df["sleep_hours"] < 5) * 1
work_risk = (df["weekly_work_hours"] > 20) * 1
attendance_risk = (df["attendance_rate"] < 60) * 2
deadlines_risk = (df["missed_deadlines"] > 3) * 1

print(type(tuition_risk))

df["risk_score"] = (
    tuition_risk
    + stress_risk
    + sleep_risk
    + work_risk
    + attendance_risk
    + deadlines_risk
)

df.groupby("risk_score").agg(
    students=("person_id", "count"),
    dropout_rate=("dropout", "mean"),
)
````

### 8. Read the GDP per capita dataset published by Our World in Data directly from the URL

```
https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true
```

and store it in a dataframe called `gdp`. The column `ny_gdp_pcap_pp_kd` contains the GDP per capita.

Clean the dataset, removing the rows that do not correspond to a single country (e.g. `World`, `High-income countries`). For each country keep only the row with the most recent `year`.

````python
gdp = pd.read_csv("https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true")
gdp.head(10)
````

````python
gdp.notna().all().all() # False means there are missing values
````

````python
gdp.isna().sum() # count of missing values per column
````

````python
# 1. Keep only rows corresponding to individual countries (non-null 3-letter ISO code)
gdp_clean = gdp[gdp["code"].notna() & (gdp["code"].str.len() == 3)].copy()
gdp_clean
````

````python
gdp_clean.isna().sum() # check for missing values again
````

````python
latest_gdp = (
gdp_clean.loc[
    gdp_clean.groupby("entity")["year"].idxmax()
    ]
    .reset_index(drop=True) # Drop=False means the index column for idxmax calculated will be shown.
)
latest_gdp
````

### 9. Join the GDP data with the students data. Compute the dropout rate for each continent.

````python
country_map = {
    "USA": "United States",
    "UK": "United Kingdom"
}

df["nationality_clean"] = df["nationality"].replace(country_map)
````

````python
students_with_gdp = df.merge(
    latest_gdp[["entity", "code", "year", "ny_gdp_pcap_pp_kd", "owid_region"]],
    left_on="nationality_clean",
    right_on="entity",
    how="left",
)

students_with_gdp.head()
````

````python
students_with_gdp.isna().sum() # check for missing values after the merge
````

````python
students_with_gdp.groupby("owid_region").agg(
    students=("person_id", "count"),        
    dropout_rate=("dropout", "mean"),
).sort_values("dropout_rate", ascending=False)
````
